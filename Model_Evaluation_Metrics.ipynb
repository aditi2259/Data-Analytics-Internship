```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# Synthetic data for classification
np.random.seed(42)
data_classification = pd.DataFrame({
'feature1': np.random.randn(100),
'feature2': np.random.randn(100),

'target': np.random.choice([0, 1], size=100)
})

# Synthetic data for regression
data_regression = pd.DataFrame({
'feature1': np.random.randn(100),
'feature2': np.random.randn(100),
'target': np.random.randn(100) * 10 + 5
})

# Split the datasets into training and testing sets
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
data_classification[['feature1', 'feature2']], data_classification['target'], test_size=0.2,
random_state=42)

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
data_regression[['feature1', 'feature2']], data_regression['target'], test_size=0.2,
random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression, LinearRegression

# Classification model
clf = LogisticRegression()

clf.fit(X_train_class, y_train_class)

# Regression model
reg = LinearRegression()
reg.fit(X_train_reg, y_train_reg)
```

```python
# Predictions for classification
y_pred_class = clf.predict(X_test_class)

# Predictions for regression
y_pred_reg = reg.predict(X_test_reg)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score

# Accuracy
accuracy = accuracy_score(y_test_class, y_pred_class)
print(f"Accuracy: {accuracy:.2f}")

# Precision
precision = precision_score(y_test_class, y_pred_class)
print(f"Precision: {precision:.2f}")

# Recall
recall = recall_score(y_test_class, y_pred_class)
print(f"Recall: {recall:.2f}")
```

```python
from sklearn.metrics import mean_squared_error

# RMSE
rmse = np.sqrt(mean_squared_error(y_test_reg, y_pred_reg))
print(f"RMSE: {rmse:.2f}")
```

```python
import matplotlib.pyplot as plt

# Plot for regression
plt.scatter(y_test_reg, y_pred_reg)
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.title('Actual vs Predicted Values')
plt.show()
```
