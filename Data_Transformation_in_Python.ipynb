```python
import pandas as pd
import numpy as np
```

```python
data = {
'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
'Age': [24, np.nan, 22, 25, 30],
'Salary': [50000, 54000, 32000, np.nan, 40000],
'Department': ['HR', 'Finance', 'IT', 'HR', 'Finance']
}

df = pd.DataFrame(data)
print(df)
```

```python

df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)

print(df)
```

```python
# Min-Max Normalization of the Salary
df['Salary'] = (df['Salary'] - df['Salary'].min()) / (df['Salary'].max() - df['Salary'].min())

print(df)
```

```python
df = pd.get_dummies(df, columns=['Department'])

print(df)
```

```python

conditions = [
(df['Age'] <= 25),
(df['Age'] > 25)
]
choices = ['Junior', 'Senior']
df['Experience_Level'] = np.select(conditions, choices, default='Unknown')

print(df)
```

```python

bins = [20, 25, 30, 35]

labels = ['20-25', '25-30', '30-35']
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

print(df)
```

```python
agg_df = df.groupby('Experience_Level').agg({'Salary': 'mean'})
print(agg_df)
```

```python
df['Salary_Rank'] = df['Salary'].rank(method='dense')

print(df)
```

```python
df.to_csv('transformed_data.csv', index=False)
```
